```python
import pandas as pd 
df = pd.read_csv('../data/Online_Retail.csv')
df["InvoiceNo"] = df["InvoiceNo"].astype(str)
```

```python
is_cancel = df["InvoiceNo"].str.startswith("C")
negative = df["Quantity"] < 0
neg_no_cancel= df[negative & ~is_cancel]
```

```python
df_clean = df[~(negative & ~is_cancel)]
df_clean.shape
```

```python
free_rows = df_clean[(df_clean['Quantity']>0)&(df_clean['UnitPrice']==0)]
free_rows.shape
free_rows.Description
free_rows.CustomerID
```
